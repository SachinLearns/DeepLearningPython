# MNIST Handwritten Digit Classification using TensorFlow Keras

## 1. Import Required Libraries

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.layers import Dense
import matplotlib.pyplot as plt
```

## 2. Load the MNIST Dataset

```python
# Load the dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
```

## 3. Visualize Sample Images from the Dataset

```python
# Visualize sample images
plt.figure(figsize=(10, 10))
for i in range(25):  # Show 25 images
    plt.subplot(5, 5, i + 1)
    plt.xticks([])
    plt.yticks([])
    plt.grid(False)
    plt.imshow(x_train[i], cmap=plt.cm.binary)
    plt.xlabel(f"Label: {y_train[i]}")
plt.tight_layout()
plt.show()
```

## 4. Normalize the Image Data

```python
# Normalize the data
x_train = x_train / 255.0
x_test = x_test / 255.0
```

## 5. Define the Neural Network Model

```python
# Define the model
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),  # Input layer: flattens the 28x28 images
    Dense(128, activation='relu'),               # Hidden Layer with 128 neurons and ReLU activation
    Dense(10, activation='softmax')              # Output Layer with 10 neurons (for 10 classes)
])
```

## 6. Compile the Model

```python
# Compile the model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

## 7. Train the Model

```python
# Train the model
model.fit(x_train, y_train, epochs=5)
```

## 8. Evaluate the Model Performance

```python
# Evaluate the model
loss, accuracy = model.evaluate(x_test, y_test, verbose=0)
print(f"Loss: {loss:.4f}")
print(f"Accuracy: {accuracy:.4f}")
```
